```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt
```

```python
run_dir = Path("outputs/egoblind_runs")
latest = sorted(run_dir.glob("*"))[-1]
latest
```

```python
frame_df = pd.read_csv(latest / "frame_metrics.csv")
summary = json.loads((latest / "run_summary.json").read_text())
frame_df.head(), summary
```

```python
plt.figure(figsize=(10,4))
frame_df["frame_total_latency_ms"].plot(title="Total Latency Over Time")
plt.show()
```

```python
component_cols=["yolo_latency_ms","depth_latency_ms","fusion_latency_ms","navigation_latency_ms","tts_latency_ms","visualization_latency_ms"]
frame_df[component_cols].mean().plot(kind="bar",title="Average Component Latency")
plt.show()
```

```python
frame_df[component_cols].quantile(0.95).plot(kind="bar",title="P95 Component Latency")
plt.show()
```

```python
frame_df["fps_instant"].plot(title="FPS Over Time")
plt.show()
```

```python
frame_df["nav_command"].value_counts().plot(kind="bar",title="Command Distribution")
plt.show()
```

```python
frame_df[["left_risk","center_risk","right_risk"]].plot(title="Risk Trends")
plt.show()
```
